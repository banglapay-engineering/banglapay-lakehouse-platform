```sql
SELECT
    kafka_topic,
    kafka_partition,
    COUNT(*) AS bronze_records,
    COUNT(DISTINCT kafka_offset) AS unique_offsets,
    MIN(kafka_offset) AS first_offset,
    MAX(kafka_offset) AS last_offset,
    SUM(
        CASE WHEN schema_id IS NULL THEN 1 ELSE 0 END
    ) AS missing_schema_ids,
    MAX(_ingested_at) AS latest_ingestion_time
FROM banglapay.bronze.kafka_events
WHERE _environment = 'dev'
GROUP BY
    kafka_topic,
    kafka_partition
ORDER BY
    kafka_topic,
    kafka_partition;
```
